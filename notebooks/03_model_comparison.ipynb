# 03 - Model Comparison

Produces the two comparisons required by the assignment:

1. **Retriever comparison** - BiomedCLIP vs ColPali (Recall@k, MRR, latency).
2. **Report-strategy comparison** - MedGemma zero-shot vs MedGemma + RAG few-shot
   (BLEU, ROUGE-L, BERTScore).

Prerequisites: built indices (`scripts/build_index.py`) and a QA dataset
(notebook 01). The ColPali index is built in notebook 02 on Colab; if it is
absent, that row is skipped gracefully.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))

import random
from cxr.config import CONFIG
from cxr.data.qa_builder import load_qa_dataset
from cxr.evaluation.compare import compare_retrievers, compare_report_strategies, _markdown_table

qa = load_qa_dataset()
qa_sample = random.Random(CONFIG.seed).sample(qa, min(CONFIG.evaluation.sample_size, len(qa)))
len(qa_sample)
```

```python
import pandas as pd

retriever_rows = compare_retrievers(qa_sample)
pd.DataFrame(retriever_rows)
```
